```python
import brie
import numpy as np
```

```python
gtf_file = 'mouse_SE.lenient_50events.gff3'

genes = brie.io.read_gff(gtf_file)
```

```python
genes[0].chrom
```

```python
genes[0].trans[0].exons
```

```python
import pandas as pd

df = pd.DataFrame(
    {'geneID': [g.geneID for g in genes],
     'geneName': [g.geneName for g in genes],
     'chrom': [g.chrom for g in genes],
     'exons': [','.join(['-'.join(e) for e in g.trans[0].exons.astype('str')]) for g in genes]
    }
)
```
